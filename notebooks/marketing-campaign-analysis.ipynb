## 2. Data Cleaning

In this section, we clean and prepare the dataset for further analysis.

The cleaning process includes handling missing values, correcting invalid records,
converting data types, and removing unnecessary columns.

```python
df_clean = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", df_clean.shape)
```

```python
missing_summary = pd.DataFrame({
    "Missing values" : df_clean.isnull().sum(),
    "Missing percentage" : (df_clean.isnull().mean()* 100).round(2)
})

missing_summary[missing_summary["Missing values"]> 0]
```

```python
income_median = df_clean["Income"].median()

df_clean["Income"] = df_clean["Income"].fillna(income_median)

print("Income median:", income_median)
print("Missing Income values:", df_clean["Income"].isnull().sum())
```

```python
invalid_birth_years = df_clean["Year_Birth"] < 1940

print("Invalid birth years:", invalid_birth_years.sum())

df_clean.loc[invalid_birth_years, "Year_Birth"] = np.nan

print("Remaining invalid birth years:", (df_clean["Year_Birth"]< 1940).sum())
```

```python
invalid_income = df_clean["Income"] == 666666

print("Invalid income values:", invalid_income.sum())

df_clean.loc[invalid_income, "Income"] = np.nan

df_clean["Income"] = df_clean["Income"].fillna(
    df_clean["Income"].median()
)

print("Remaining missing Income:")
df_clean["Income"].isnull().sum()
```

```python
df_clean["Dt_Customer"] = pd.to_datetime(
    df_clean["Dt_Customer"],
    format="%d-%m-%Y",
    errors="coerce"
)

print(df_clean["Dt_Customer"].dtype)
print("Missing dates:", df_clean["Dt_Customer"].isnull().sum())
```

```python
print("Z_CostContact unique values:", df_clean["Z_CostContact"].unique())

print("Z_Revenue unique values: ", df_clean["Z_Revenue"].unique())
```

```python
df_clean = df_clean.drop(
    columns=["Z_CostContact", "Z_Revenue"]
)

print("shape after dropping constant columns :", df_clean.shape)
```

```python
print("Dataset shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:", df_clean.duplicated().sum())

print("\nData types:")
print(df_clean.dtypes)

print("\nDataset information")
df_clean.info()
```

## 3. Feature Engineering

In this section, we create new features from existing customer data
to support customer segmentation and marketing campaign analysis.

The new features include customer age, household composition,
total spending, purchase behavior, and customer tenure.


```python
reference_year = 2014

df_clean["Age"] = reference_year - df_clean["Year_Birth"]

df_clean[["Year_Birth", "Age"]].head()
```

```python
df_clean["Age"].describe()
```

```python
df_clean["Children"] = (
    df_clean["Kidhome"] + df_clean["Teenhome"]
)

df_clean[["Kidhome", "Teenhome", "Children"]].head()
```

```python
df_clean["Has_Children"] = (
    df_clean["Children"] > 0
).astype(int)

df_clean["Has_Children"].value_counts()
```

```python
spending_columns = [
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

df_clean["Total_Spending"] = df_clean[spending_columns].sum(axis=1)

df_clean[["Total_Spending"]+ spending_columns].head()
```

```python
df_clean["Total_Spending"].describe()
```

```python
purchase_columns = [
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases"
]

df_clean["Total_Purchases"] = df_clean[purchase_columns].sum(axis=1)

df_clean[["Total_Purchases"] + purchase_columns].head()
```

```python
reference_date = df_clean["Dt_Customer"].max()

print("Reference date:", reference_date)


df_clean["Customer_Tenure_Days"] = (
    reference_date - df_clean["Dt_Customer"]
).dt.days

df_clean["Customer_Tenure_Days"].describe()
```

```python
print(
    "Negative tenure values:",
    (df_clean["Customer_Tenure_Days"]<0).sum()
)
```

```python
age_bins = [0, 30, 45, 60, 100]
age_labels = [
    "Under 30",
    "30-44",
    "45-59",
    "60+"
]

df_clean["Age_Group"] = pd.cut(
    df_clean["Age"],
    bins = age_bins,
    labels= age_labels,
    right= False
)

df_clean["Age_Group"].value_counts(dropna=False)
```

```python
df_clean["Spending_Group"] = pd.qcut(
    df_clean["Total_Spending"],
    q=3,
    labels=["Low Spender", "Medium Spender", "High Spender"]
)

print(df_clean["Spending_Group"].value_counts())
```

```python
new_columns = [
    "Age",
    "Children",
    "Has_Children",
    "Total_Spending",
    "Total_Purchases",
    "Customer_Tenure_Days",
    "Age_Group",
    "Spending_Group"
]

print("Dataset shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean[new_columns].isnull().sum())

print("\nNegative tenure:")
print((df_clean["Customer_Tenure_Days"] < 0).sum())

print("\nAge groups:")
print(df_clean["Age_Group"].value_counts(dropna=False))

print("\nSpending groups:")
print(df_clean["Spending_Group"].value_counts())

print("\nNew features:")
display(df_clean[new_columns].head())
```

### Feature Engineering - Summary

Several new features were created to support customer behavior analysis:

- Age: Customer age calculated from birth year.
- Children: Total number of children and teenagers in the household.
- Has_Children: Indicates whether a customer has children.
- Total_Spending: Total spending across all product categories.
- Total_Purchases: Total purchases across different shopping channels.
- Customer_Tenure_Days: Number of days since customer registration.
- Age_Group: Customer segmentation based on age.
- Spending_Group: Customer segmentation based on total spending.

These features will be used in the exploratory data analysis stage.

## 5. SQL Analysis

In this section, SQL is used to analyze customer data stored in a SQLite database.

The analysis focuses on customer statistics, spending behavior,
product preferences, shopping channels, and marketing campaign performance.

SQL queries are executed through Python using SQLAlchemy and Pandas.

```python
from sqlalchemy import create_engine
```

```python
engine = create_engine(
    "sqlite:///../data/marketing_campaign.db"
)
```

```python
df_sql = df_clean.copy()

df_sql["Age_Group"] = df_sql["Age_Group"].astype("string")

df_sql["Spending_Group"] = df_sql["Spending_Group"].astype("string")
```

```python
df_sql.to_sql(
    name="marketing_campaign",
    con=engine,
    if_exists="replace",
    index=False
)

print("Data successfully saved to SQLite database.")
```

```python
tables = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type = 'table';
    """,
    engine
)
display(tables)
```

```python
query = """
SELECT
    COUNT(*) AS total_customers,
    ROUND(AVG(Age), 2) AS average_age,
    ROUND(AVG(Income), 2) AS average_income,
    ROUND(AVG(Total_Spending), 2) AS average_spending,
    ROUND(AVG(Total_Purchases), 2) AS average_purchases
FROM marketing_campaign;
"""

result = pd.read_sql_query(query, engine)

display(result)
```

```python
query = """
SELECT
    'Wines' AS product_category,
    SUM(MntWines) AS total_spending
FROM marketing_campaign

UNION ALL

SELECT
    'Meat Products',
    SUM(MntMeatProducts)
FROM marketing_campaign

UNION ALL

SELECT
    'Gold Products',
    SUM(MntGoldProds)
FROM marketing_campaign

UNION ALL

SELECT
    'Fish Products',
    SUM(MntFishProducts)
FROM marketing_campaign

UNION ALL

SELECT
    'Sweet Products',
    SUM(MntSweetProducts)
FROM marketing_campaign

UNION ALL

SELECT
    'Fruits',
    SUM(MntFruits)
FROM marketing_campaign

ORDER BY total_spending DESC;
"""

product_sql = pd.read_sql_query(query, engine)

display(product_sql)
```

```python
query = """
SELECT
    'Web' AS purchase_channel,
    SUM(NumWebPurchases) AS total_purchases
FROM marketing_campaign

UNION ALL

SELECT
    'Catalog',
    SUM(NumCatalogPurchases)
FROM marketing_campaign

UNION ALL

SELECT
    'Store',
    SUM(NumStorePurchases)
FROM marketing_campaign

ORDER BY total_purchases DESC;
"""

channel_sql = pd.read_sql_query(query, engine)

display(channel_sql)
```

```python
query = """
SELECT
    'Campaign 1' AS campaign,
    ROUND(AVG(AcceptedCmp1) * 100, 2) AS acceptance_rate
FROM marketing_campaign

UNION ALL

SELECT
    'Campaign 2',
    ROUND(AVG(AcceptedCmp2) * 100, 2)
FROM marketing_campaign

UNION ALL

SELECT
    'Campaign 3',
    ROUND(AVG(AcceptedCmp3) * 100, 2)
FROM marketing_campaign

UNION ALL

SELECT
    'Campaign 4',
    ROUND(AVG(AcceptedCmp4) * 100, 2)
FROM marketing_campaign

UNION ALL

SELECT
    'Campaign 5',
    ROUND(AVG(AcceptedCmp5) * 100, 2)
FROM marketing_campaign

UNION ALL

SELECT
    'Last Campaign',
    ROUND(AVG(Response) * 100, 2)
FROM marketing_campaign

ORDER BY acceptance_rate DESC;
"""

campaign_sql = pd.read_sql_query(query, engine)

display(campaign_sql)
```

```python
query = """
SELECT
    Spending_Group,
    COUNT(*) AS customer_count,
    ROUND(AVG(Total_Spending), 2) AS average_spending,
    ROUND(SUM(Total_Spending), 2) AS total_spending
FROM marketing_campaign
GROUP BY Spending_Group
ORDER BY
    CASE Spending_Group
        WHEN 'Low Spender' THEN 1
        WHEN 'Medium Spender' THEN 2
        WHEN 'High Spender' THEN 3
    END;
"""

spending_sql = pd.read_sql_query(query, engine)

display(spending_sql)
```

```python
query = """
SELECT
    CASE
        WHEN Has_Children = 1 THEN 'Has Children'
        ELSE 'No Children'
    END AS family_status,
    COUNT(*) AS customer_count,
    ROUND(AVG(Total_Spending), 2) AS average_spending,
    ROUND(AVG(Income), 2) AS average_income
FROM marketing_campaign
GROUP BY Has_Children;
"""

family_sql = pd.read_sql_query(query, engine)

display(family_sql)
```

```python
query = """
SELECT
    Age_Group,
    COUNT(*) AS customer_count,
    ROUND(AVG(Income), 2) AS average_income,
    ROUND(AVG(Total_Spending), 2) AS average_spending
FROM marketing_campaign
WHERE Age_Group IS NOT NULL
GROUP BY Age_Group
ORDER BY
    CASE Age_Group
        WHEN 'Under 30' THEN 1
        WHEN '30-44' THEN 2
        WHEN '45-59' THEN 3
        WHEN '60+' THEN 4
    END;
"""

age_sql = pd.read_sql_query(query, engine)

display(age_sql)
```

```python
query = """
SELECT
    COUNT(*) AS total_records,
    COUNT(DISTINCT ID) AS unique_customers
FROM marketing_campaign;
"""

database_check = pd.read_sql_query(query, engine)

display(database_check)

print("DataFrame rows:", len(df_clean))
print("Database rows:", database_check.loc[0, "total_records"])
```

## 6. Business Insights and Recommendations

This section translates the findings from exploratory data analysis
and SQL analysis into actionable business insights.

The objective is to identify valuable customer segments,
understand purchasing behavior, evaluate marketing campaign performance,
and provide data-driven recommendations.

```python
query = """
SELECT
    Age_Group,
    COUNT(*) AS customer_count,
    SUM(Response) AS accepted_customers,
    ROUND(AVG(Response) * 100, 2) AS acceptance_rate
FROM marketing_campaign
WHERE Age_Group IS NOT NULL
GROUP BY Age_Group
ORDER BY acceptance_rate DESC;
"""

age_campaign_analysis = pd.read_sql_query(query, engine)

display(age_campaign_analysis)
```

```python
query = """
SELECT
    CASE
        WHEN Has_Children = 1 THEN 'Has Children'
        ELSE 'No Children'
    END AS family_status,
    COUNT(*) AS customer_count,
    SUM(Response) AS accepted_customers,
    ROUND(AVG(Response) * 100, 2) AS acceptance_rate,
    ROUND(AVG(Total_Spending), 2) AS average_spending
FROM marketing_campaign
GROUP BY Has_Children;
"""

family_campaign_analysis = pd.read_sql_query(query, engine)

display(family_campaign_analysis)
```

```python
query = """
SELECT
    Spending_Group,
    COUNT(*) AS customer_count,
    SUM(Response) AS accepted_customers,
    ROUND(AVG(Response) * 100, 2) AS acceptance_rate,
    ROUND(AVG(Income), 2) AS average_income,
    ROUND(AVG(Total_Spending), 2) AS average_spending
FROM marketing_campaign
GROUP BY Spending_Group
ORDER BY acceptance_rate DESC;
"""

spending_campaign_analysis = pd.read_sql_query(query, engine)

display(spending_campaign_analysis)
```

```python
query = """
SELECT
    COUNT(*) AS customer_count,
    ROUND(AVG(Income), 2) AS average_income,
    ROUND(AVG(Total_Spending), 2) AS average_spending,
    ROUND(AVG(Response) * 100, 2) AS campaign_acceptance_rate
FROM marketing_campaign
WHERE Income >= (
    SELECT AVG(Income)
    FROM marketing_campaign
)
AND Total_Spending >= (
    SELECT AVG(Total_Spending)
    FROM marketing_campaign
);
"""

valuable_customers = pd.read_sql_query(query, engine)

display(valuable_customers)
```
